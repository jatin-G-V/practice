```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("cleaned_data.csv")

df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
# Creating a histogram of Salary

plt.hist(df["Salary"], bins=8)

plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.title("Salary Distribution")

plt.show()
```

```python
# Creating a boxplot of Salary

plt.boxplot(df["Salary"])

plt.ylabel("Salary")
plt.title("Salary Boxplot")

plt.show()
```

```python
# Creating a scatter plot of Age vs Salary

plt.scatter(df["Age"], df["Salary"])

plt.xlabel("Age")
plt.ylabel("Salary")
plt.title("Age vs Salary")

plt.show()
```

```python
# Selecting numeric columns for correlation

numeric_df = df[["Age", "Salary"]]

correlation_matrix = numeric_df.corr()

print(correlation_matrix)
```

```python
# Creating a correlation heatmap

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```
