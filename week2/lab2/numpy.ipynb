# 2D NumPy Array

```python
# Creating a 2D Array
import numpy as np

data = np.array([
    [10, 20, 30],
    [20, 40, 60],
    [30, 50, 90],
    [40, 60, 120]
])

print(data)
```

```python
# Calculating the mean of each row
import numpy as np

data = np.array([
    [10, 20, 30],
    [20, 40, 60],
    [30, 50, 90],
    [40, 60, 120]
])

print(data)
```

```python
# Calucalting the maximum value of each column
import numpy as np

column_maxima = data.max(axis=0)

print("Column maxima:", column_maxima)
```

# Normalize each column to 0–1

```python
# Normalizing the data of each column in the range of 0 to 1
column_min = data.min(axis=0)
column_max = data.max(axis=0)

normalized_data = (data - column_min) / (column_max - column_min)
print(data)
print("Normalized data:")
print(normalized_data)
```

# Broadcasting

```python
# Broadcasting allows NumPy to perform operations between arrays of compatible shapes without explicitly repeating the smaller array.
column_means = data.mean(axis=0)

print("Column means:", column_means)

centered_data = data - column_means

print("Centered data:")
print(centered_data)
```

# Vectorization

```python
# Creating a large Array
import numpy as np
import time

data = np.random.rand(1000000)
```

```python
# Calculating the time of processing each element with loop
start = time.time()

loop_result = []

for value in data:
    loop_result.append(value * 2)

loop_time = time.time() - start

print("Loop time:", loop_time)
```

```python
# With Vectorized numpy operation
start = time.time()

vectorized_result = data * 2

vectorized_time = time.time() - start

print("Vectorized time:", vectorized_time)
```
