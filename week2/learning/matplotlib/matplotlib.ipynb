```python
import matplotlib.pyplot as plt
```

```python
# Ploting a graph using x and y values

x = [1,2,3,4,5]
y = [2,4,6,8,10]
plt.plot(x,y)
plt.show()
```

```python
# Adding labels and title to the line plot

x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x, y)

plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.title("Relationship between X and Y")

plt.show()
```

```python
# Customizing the line plot using markers, line style and color

x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x, y, marker="o", linestyle="--", color="red")

plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.title("Relationship between X and Y")

plt.show()
```

```python
# Setting figure size and adding grid to the plot

x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.figure(figsize=(8, 5))

plt.plot(x, y, marker="o")

plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.title("Relationship between X and Y")

plt.grid(True)

plt.show()
```

```python
# Setting figure size and adding grid to the plot

x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.figure(figsize=(8, 5))

plt.plot(x, y, marker="o")

plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.title("Relationship between X and Y")

plt.grid(False)

plt.show()
```

```python
# Plotting multiple lines on the same graph

x = [1, 2, 3, 4, 5]

y1 = [2, 4, 6, 8, 10]
y2 = [1, 3, 5, 7, 9]

plt.plot(x, y1, marker="o", color = "red", linestyle = "--", label="Line 1")
plt.plot(x, y2, marker="s", color = "blue",label="Line 2")

plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.title("Comparison of Two Lines")

plt.legend()
plt.grid(True)

plt.show()
```

```python
# Creating multiple plots in a single figure

x = [1, 2, 3, 4, 5]

y1 = [2, 4, 6, 8, 10]
y2 = [10, 8, 6, 4, 2]

plt.subplot(2, 2, 1) 

# (2,2,1) : (2 - refers to number of rows in grid,2 - refers to number of columns in grid , 1 refers the postion of graph in the grid)

plt.plot(x, y1, marker="o")
plt.title("Increasing")

plt.subplot(2, 2, 2)
plt.plot(x, y2, marker="s")
plt.title("Decreasing")


plt.subplot(2, 2, 3)
plt.plot(x, y2, marker="s")
plt.title("Decreasing")

plt.tight_layout()

plt.show()
```

```python
# Creating a basic bar chart

categories = ["Python", "SQL", "Java", "C++"]
values = [10, 25, 15, 30]

plt.bar(categories, values)

plt.xlabel("Categories")
plt.ylabel("Values")
plt.title("Category-wise Values")

plt.show()
```

```python
# Creating a horizontal bar chart

categories = ["Python", "SQL", "Java", "C++"]
values = [90, 80, 70, 60]

plt.barh(categories, values)

plt.xlabel("Score")
plt.ylabel("Skills")
plt.title("Skill-wise Score")

plt.show()
```

```python
# Creating a scatter plot

x = [1, 2, 3, 4, 5, 6]
y = [2, 4, 5, 8, 9, 12]

plt.scatter(x, y)

plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.title("Relationship between X and Y")

plt.show()
```

```python
# Creating a histogram

data = [10, 12, 12, 14, 14, 14, 15, 16, 17,
        18, 18, 19, 20, 21, 22, 22, 23, 24, 25]

plt.hist(data,bins = 1)

plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Distribution of Values")

plt.show()
```

```python
# Creating a pie chart

categories = ["Python", "SQL", "Java", "C++"]
values = [40, 30, 20, 10]

plt.pie(values, labels=categories,autopct="%1.1f")

plt.title("Skill Distribution")

plt.show()
```

```python
# Creating a boxplot

data = [10, 12, 13, 14, 15, 15, 16, 17, 18, 20, 22, 25, 30]

plt.boxplot(data)

plt.ylabel("Values")
plt.title("Distribution of Data")

plt.show()
```

```python
# Creating a boxplot with data having a outlier

data = [10, 12, 13, 14, 15, 15, 16, 17, 18, 20, 22, 25, 30, 100]

plt.boxplot(data)

plt.ylabel("Values")
plt.title("Distribution of Data")

plt.show()
```

```python
# Comparing multiple dataset with boxplot
data1 = [10, 12, 13, 14, 15, 16, 17]
data2 = [20, 22, 25, 28, 30, 35, 40]

plt.boxplot([data1, data2], tick_labels=["Group A", "Group B"])

plt.ylabel("Values")
plt.title("Comparison of Two Groups")

plt.show()
```

```python
#Learning concepts :- xlim(), ylim(), xticks(), yticks()

# Setting limits for x-axis and y-axis

x = [1, 2, 3, 4, 5,6]
y = [2, 4, 6, 7, 8, 10]
plt.plot(x, y, marker="o")

plt.xlim(0, 6)
plt.ylim(0, 12)

plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.title("Setting Axis Limits")

plt.show()
```

```python
# Customizing tick positions on the axes

plt.plot(x, y, marker="o")

plt.xticks([1, 2, 3, 4, 5, 6])
plt.yticks([2, 4, 6, 7, 8, 10])

plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.title("Custom Axis Ticks")

plt.show()
```

```python
x = [1, 2, 3, 4]
y = [80, 90, 70, 95]

plt.plot(x, y, marker="o")

plt.xticks(
    [1, 2, 3, 4],
    ["Python", "SQL", "Java", "C++"]
)

plt.ylabel("Score")
plt.title("Skill Scores")

plt.show()
```

```python
# Creating a figure and axes using subplots

x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

fig, ax = plt.subplots()

ax.plot(x, y, marker="o")

ax.set_xlabel("X Values")
ax.set_ylabel("Y Values")
ax.set_title("Relationship between X and Y")

ax.grid(True)

plt.show()
```

```python
# Creating multiple plots using subplots and axes

x = [1, 2, 3, 4, 5]

y1 = [2, 4, 6, 8, 10]
y2 = [10, 8, 6, 4, 2]

fig, axes = plt.subplots(2, 2, figsize=(10, 8))

axes[0,0].plot(x, y1, marker="o")
axes[0,0].set_title("Increasing")

axes[1,1].plot(x, y2, marker="s")
axes[1,1].set_title("Decreasing")

plt.tight_layout()

fig.savefig("multiple_plot.png")

plt.show()
```

```python
import matplotlib.pyplot as plt

students = ["Aman", "Rahul", "Jatin", "Priya", "Neha"]
marks = [65, 80, 72, 90, 55]

fig, ax = plt.subplots()

ax.bar(students, marks)

ax.set_xlabel("Students")
ax.set_ylabel("Marks")
ax.set_title("Class Marks Report")

ax.grid(True)

fig.savefig("student_marks.png")

plt.show()
```
