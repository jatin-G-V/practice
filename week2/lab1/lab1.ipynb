# Loading a csv into DataFrame

```python
import pandas as pd

df = pd.read_csv("WineQT.csv")

print(df.head())
print(df.info())
print(df.describe())
print(df.shape)
```

### 

```python
# print all rows with columns alcohol and quality
df.loc[:, ["alcohol", "quality"]]


```

```python
# print five rows with columns alcohol and quality
df.loc[0:4, ["alcohol", "quality"]]
```

```python
# Retrieving data with index
# First 5 rows aur first 3 columns:
df.iloc[0:5, 0:3]
```

```python
# First row
df.iloc[0]
```

# Boolean Filtering

```python
# Print the data where alcohol > 12
df[df["alcohol"] > 12]
```

```python
# Multiple Condition (AND)
df[(df["alcohol"] > 12) & (df["quality"] >= 7)]
```

```python
# Multiple Condition (OR)
df[(df["alcohol"] > 13) | (df["quality"] >= 8)]
```

```python
# Data of alcohol and quality where alcohol > 12
df.loc[df["alcohol"] > 12, ["alcohol", "quality"]]
```

# Creating new columns with existing columns

```python
# Creating alcohol_density_ratio column with alcohol and density column
df["alcohol_density_ratio"] = df["alcohol"] / df["density"]
```

```python
df[["alcohol", "density", "alcohol_density_ratio"]].head()
```

# GroupBY() + Aggregation

```python
#. groupby() + mean()
print("Average alcohol of each quality level")
df.groupby("quality")["alcohol"].mean()
```

```python
# groupby() + count()
df.groupby("quality")["alcohol"].count()
```

```python
# groupby()  + mean + count
df.groupby("quality")["alcohol"].agg(["mean", "count"])
```

```python
quality_summary = df.groupby("quality")["alcohol"].agg(
    mean_alcohol="mean",
    wine_count="count"
)

print(quality_summary)
```

# Merge()

```python
# Creating two DataFrames
df1 = df[["Id", "quality"]].copy()

df2 = df[["Id", "alcohol"]].copy()
```

```python
# Merging the DataFrames
merged_df = pd.merge(df1, df2, on="Id")
print(merged_df.head())
```

```python
# Merging and comparing the row count
df1 = df[["Id", "quality"]].copy()
df2 = df[["Id", "alcohol"]].copy()

merged_df = pd.merge(df1, df2, on="Id")

print(merged_df.head())
print("Original rows:", len(df))
print("Merged rows:", len(merged_df))
```
